# 本番環境への移行計画

## 目的
- 現在のローカル環境を本番環境（AWS)に展開する計画を策定する
- 環境変数・シークレット管理の設計
- ログ・モニタリングの設計
- 障害時の対応フローを定義

## 現状と目標

| 項目 | 現状（ローカル） | 目標（本番） |
|---|---|---|
| データベース | SQLite（ファイル） | PostgreSQL or BigQuery |
| 実行環境 | ローカル PC | AWS EC2 or Lambda |
| スケジュール | 手動実行 | Airflow で毎日自動 |
| ストレージ | ローカルファイル | AWS S3 |
| 監視 | なし | CloudWatch + アラート |
| テスト | ローカル pytest | GitHub Actions（実装済み） |
| シークレット | なし | AWS Secrets Manager |

```python
# 本番環境の設定テンプレート
import json
from datetime import datetime

# 本番環境の設定を定義
production_config = {
    "project": "nikkei225-analysis",
    "version": "1.0.0",
    "last_updated": datetime.now().strftime("%Y-%m-%d"),
    
    "environments": {
        "development": {
            "database": "SQLite (nikkei225.db)",
            "execution": "Local PC",
            "storage": "Local filesystem",
            "scheduler": "Manual",
            "monitoring": "None"
        },
        "staging": {
            "database": "PostgreSQL (AWS RDS)",
            "execution": "AWS EC2 (t3.micro)",
            "storage": "AWS S3 (staging bucket)",
            "scheduler": "Airflow (Docker on EC2)",
            "monitoring": "CloudWatch (basic)"
        },
        "production": {
            "database": "PostgreSQL (AWS RDS) or BigQuery",
            "execution": "AWS Lambda or ECS",
            "storage": "AWS S3 (production bucket)",
            "scheduler": "Airflow (managed) or Step Functions",
            "monitoring": "CloudWatch + SNS alerts"
        }
    }
}

print("【環境設定】\n")
for env_name, config in production_config["environments"].items():
    print(f"═══ {env_name.upper()} ═══")
    for key, value in config.items():
        print(f"  {key:<15}: {value}")
    print()
```
